```python
def fact(n):
    try:
        if n < 0:
            raise ValueError("Invalide input")
        if n == 0 or n == 1:
            return n
        return n*fact(n-1)
    except ValueError as e:
        print(e)

fact(-1)
```

```python
'''Write a generator function that yields Fibonacci numbers up to a given limit n.'''


def fibo(n):
    a = 0
    b = 1
    for _ in range(n):
        a,b = b,a+b
        yield a


y = fibo(11)
for i in y:
    print(i)
```

```python
'''Write a generator function that yields Fibonacci numbers up to a given limit n.'''


def fibo(n):
    a = 0
    b = 1
    while n:
        # if a > n: break
        yield a
        a,b = b,a+b
        n -= 1


y = fibo(11)
for i in y:
    print(i)
```

```python
'''  Implement a stack data structure using a Python class with push, pop, peek, and is_empty methods.'''


class stack:
    def __init__(self):
        self.s = []
    
    def push(self,num):
        self.s.append(num)
        return self.s
    
    def pop(self):
        return self.s.pop()
    
    def peek(self):
        return self.s[-1]
    
    def isEmpty(self):
        return True if len(self.s) == 0 else False
    
    def __str__(self):
        return str(self.s)
    
st = stack()

print(st.push(10))
print(st.push(9))
print(st.push(8))
print(st.push(32))


print(st.pop())
print(st)
print(st.isEmpty())

print(st.peek())
```

```python
'''
Design a class hierarchy: Shape (base) → Circle and Rectangle (subclasses). Implement area() and perimeter() polymorphically.
 Add a class method to track total shapes created.
'''


class shape:
    counter = 0
    def __init__(self):
        shape.counter += 1

    @classmethod
    def giveCount(cls):
        return cls.counter
    
class Circle(shape):
    def __init__(self,r):
        super().__init__()
        self.r = r
    
    def area(self):
        return 3.14*self.r*self.r
    
    def perimeter(self):
        return 2*3.14*self.r
    
class Rectangle(shape):
    def __init__(self,b,l):
        super().__init__()
        self.b = b
        self.l = l

    def area(self):
        return self.b*self.l
    
    def perimeter(self):
        return 2*(self.l + self.b)
    

C1 = Circle(10)
C2 = Circle(12.2)

print(shape.giveCount())

R1 = Rectangle(12,30)
R2 = Rectangle(10,24.12)

print(shape.giveCount())

print(C1.area())
print(C2.area())
print(C1.perimeter())
print(C2.perimeter())


print(R1.area())
print(R2.area())
print(R1.perimeter())
print(R2.perimeter())
```

```python
'''

Demonstrate collections.defaultdict by building an inverted index: 
given a list of (word, document) tuples, create a dict mapping each word to the list of documents it appears in.

'''
from collections import defaultdict

d = defaultdict(list)

input = [
    ("apple", "doc1"),
    ("banana", "doc1"),
    ("apple", "doc2"),
    ("banana", "doc3")
]

for i in input:
    d[i[0]].append(i[1])

print(d)
```
