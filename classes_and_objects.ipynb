a class is an executable object that creates a new type.


A class is an object whose job is to create other objects, and to define:

- how those objects store data
- how attribute lookup works
- what methods are available
- how instances behave when used with operators or built-ins

An instance is a dynamically allocated object that:
- has its own storage (__dict__)
- is associated with a class
- delegates behavior to that class

Important:
- Instances do not contain methods
- Methods live on the class
- The instance only stores data

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self.balance=balance

    def deposit(self,amount):
        self.balance+=amount
        logger.info(f"{amount} is deposited. New balance is {self.balance}")

    def withdraw(self,amount):
        if amount>self.balance:
            logger.info("Insufficient funds!")
        else:
            self.balance-=amount
            logger.info(f"{amount} is withdrawn. New Balance is {self.balance}")

    def get_balance(self):
        return self.balance
```

```python
account=BankAccount("Sahil",5000)
logger.info(account.balance)
account.deposit(100)
account.withdraw(300)
```
