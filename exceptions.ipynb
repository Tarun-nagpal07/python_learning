```python
from loguru import logger
```

```python
try:
    a = b
except NameError as ex:
    logger.error(ex)
```

```python
try:
    result = 1 / 0
except ZeroDivisionError as ex:
    logger.info(ex)
    logger.info("Div cannot be zero!!")
```

```python
try:
    num = "python"
    result = 10 / num
except ValueError:
    logger.error("That's not a valid number!")
except ZeroDivisionError:
    logger.error("You can't divide by zero!")
except Exception as ex:
    logger.error(ex)
else:
    logger.info(f"the result is {result}")
finally:
    logger.info("Execution complete.")
```
