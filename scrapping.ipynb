```python
from bs4 import BeautifulSoup
from loguru import logger
import requests
```

```python
with open("website.html", 'r') as file:
    content = file.read()


soup = BeautifulSoup(content, 'html.parser') # maybe also can write lxml
logger.info(soup)
```

```python
logger.info(soup.prettify())
```

```python
logger.info(soup.title)
logger.info(soup.title.name)
logger.info(soup.title.string)
```

```python
logger.info(soup.a)
logger.info(soup.a.name)
logger.info(soup.a.string)
```

```python
logger.info(soup.li)
logger.info(soup.li.name)
logger.info(soup.li.string)
```

```python
all_anchor_tags = soup.find_all(name='a')
logger.info(all_anchor_tags)
```

```python
for tags in all_anchor_tags:
    logger.info(tags.getText())
    logger.info(tags.get("href"))
```

```python
heading = soup.find(name="h1", id="name")
logger.info(heading)
```

```python
section_heading = soup.find(name="h3", class_="heading")
logger.info(section_heading)
logger.info(section_heading.name)
logger.info(section_heading.get("class"))
logger.info(section_heading.getText())
```

```python
company_url = soup.select_one(selector="p a")
logger.info(company_url)
logger.info(company_url.name)
```

```python
name = soup.select_one(selector="#name")
logger.info(name)
```

```python
headings = soup.select(".heading") # get all class = heading
logger.info(headings)
```

# Live Website

```python
response = requests.get("https://news.ycombinator.com/news")
logger.info(response)
```

```python
yc_web_page = response.text
soup = BeautifulSoup(yc_web_page, "html.parser")
logger.info(soup.title)
```

```python
news_headings = soup.select(".titleline > a")

for heading in news_headings:
    logger.info(heading.get_text(strip=True))
```

```python
news_link = soup.select(".titleline > a")

for links in news_link:
    logger.info(links.get("href"))
```

```python
news_upvotes = soup.find_all(name="span", class_="score")
for votes in news_upvotes:
    logger.info(votes.getText())
```
