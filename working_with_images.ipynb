```python
from PIL import Image
```

```python
img = Image.open('image.png')
print(img.format)
print(img.size)
print(img.mode)
img.show()
```

Resize Image

```python
resized = img.resize((500,300))
# resized.save("resized.png")
resized.show()
```

Crop Image

```python
crop_img = img.crop((50,50,300,300))
crop_img.show()
```

Rotate Image

```python
rotated = img.rotate(300)
rotated.show()
```

Convert Image Formate

```python
img = Image.open("image.png")
img.save("output.jpg","JPEG")
```

Add Text/ Watermark

```python
from PIL import Image , ImageDraw, ImageFont

img  = Image.open("output.jpg")
draw = ImageDraw.Draw(img)

font = ImageFont.load_default()

draw.text((50,50),"Watermark",fill="black",font=font)
img.show()
```

Filters

```python
from PIL import Image,ImageFilter

img = Image.open("output.jpg")
blur = img.filter(ImageFilter.BLUR)
blur.show()

# DETAIL
# EDGE_ENHANCE
# FIND_EDGES
# EMBOSS
# SHARPEN
```

```python
from PIL import Image , ImageEnhance

enhance = ImageEnhance.Brightness(img)
bright = enhance.enhance(1.5)
bright.show()

contrast = ImageEnhance.Contrast(img).enhance(2)
contrast.show()

sharp = ImageEnhance.Sharpness(img).enhance(3)
sharp.show()
```

Thumbnail

```python
img = Image.open('image.png')
img.thumbnail((150,150))
img.show()
```

Converting Image Mode
"RGB" - > Normal Color
"RGBA" -> Color + Transparency
"L" -> Gray

```python
gray = img.convert("L")
gray.show()
```

Flip Image

```python
horizontal = img.transpose(Image.FLIP_LEFT_RIGHT)
horizontal.show()
```
